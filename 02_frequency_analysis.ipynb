```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

vocabulary_occurrences = pd.read_csv("./QuranTotalMergedDatabase.csv")
```

```python
vocabulary_occurrences
```

```python
vocabulary_occurrences.groupby("Vocabulary ID")

print(
    vocabulary_occurrences["Vocabulary ID"].nunique()
)

(
    vocabulary_occurrences
    .groupby("Vocabulary ID")["POS"]
    .first()
    .value_counts()
)
```

```python
frequency = (
    vocabulary_occurrences
    .groupby("Vocabulary ID")
    .size()
    .sort_values(ascending=False)
)

frequency.describe()
```

```python
print("Occurs once:", (frequency == 1).sum())
print("Occurs 2-5 times:", frequency.between(2, 5).sum())
print("Occurs 6-10 times:", frequency.between(6, 10).sum())
print("Occurs >10 times:", (frequency > 10).sum())
```

```python
frequency
```

```python
# frequency.plot(kind='hist', logy=True, edgecolor='white', bins=100)

fig, ax = plt.subplots(1,1, layout="constrained")

ax.hist(frequency, 100)
ax.set_yscale("log")

ax.set_xlabel("Occurrences per Lexeme")
ax.set_ylabel("Number of Lexemes")
ax.set_title("Distribution of Lexeme Frequencies in the Qur'an")
```

```python
fig, ax = plt.subplots()
ax.plot(
    np.arange(1, len(frequency) + 1),
    frequency.values
)
ax.set_yscale('log')
# ax.set_xscale('log')

# ax.set_xticks([0, frequency.shape[0] - 1], ["1", str(frequency.shape[0])])
ax.set_xlabel("Frequency Rank (1 = most frequent)")
ax.set_ylabel("Number of Occurrences")
ax.set_title("Rank-Frequency Distribution of Quranic Lexemes")
```

```python
frequency = frequency.sort_values(ascending=False)

cumulative_coverage = (
    frequency.cumsum()
    / frequency.sum()
    * 100
)

for target in [50, 75, 80, 85, 90, 95, 99]:
    number_needed = (
        np.searchsorted(
            cumulative_coverage.to_numpy(),
            target,
            side="left"
        )
        + 1
    )

    print(f"{target}%: {number_needed} lexemes")
```
